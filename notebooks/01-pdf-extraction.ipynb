```python
from pypdf import PdfReader

# Define the location of the PDF file
pdf_path = "/kaggle/input/datasets/makhosazanemthethwa/cos802-university-documents/General_Academic_Regulations_2026.pdf"

# Load the PDF document
reader = PdfReader(pdf_path)

# Display the total number of pages to confirm the PDF loaded correctly
print(f"Pages: {len(reader.pages)}")

# Extract text from the first page for inspection
text = reader.pages[0].extract_text()

# Display the first 2000 characters of extracted text
print(text[:2000])
```

```python
# Create an empty string to store text from all pages
full_text = ""

# Loop through every page in the PDF
for page in reader.pages:
    
    # Extract text from the current page
    page_text = page.extract_text()

    # Only append pages that contain text
    if page_text:
        full_text += page_text + "\n"

# Display basic statistics about the extracted document
print("Characters:", len(full_text))
print("Words:", len(full_text.split()))
```

```python
# Save the extracted text to a file for future processing
with open("general_academic_regulations.txt", "w", encoding="utf-8") as f:
    f.write(full_text)

print("Text file saved successfully")
```

```python
# Display a larger sample of extracted text to assess quality
print(full_text[:5000])
```

```python
# Identify pages where text extraction may have failed
empty_pages = []

for page_number, page in enumerate(reader.pages):

    page_text = page.extract_text()

    if not page_text:
        empty_pages.append(page_number + 1)

print(f"Empty Pages: {empty_pages}")
```

```python
# Summary of extraction results
print("PDF Extraction Complete")
print(f"Pages Processed: {len(reader.pages)}")
print(f"Characters Extracted: {len(full_text):,}")
print(f"Words Extracted: {len(full_text.split()):,}")
```

## Observations
The General Academic Regulations and Student Rules 2026 document was successfully processed using the pypdf library.

A total of 123 pages were extracted, resulting in approximately 361,882 characters and 53,391 words of machine-readable text.

These results indicate that the document can be used directly in the RAG pipeline without requiring Optical Character Recognition (OCR). The size of the document also demonstrates the need for chunking and retrieval mechanisms, as the entire document cannot be supplied to a language model as context during question answering.
